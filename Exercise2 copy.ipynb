```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Parameters
L_v = 120           # bytes (video)
L_d = 135           # bytes (data)
C_link = 1e6        # bits/s (1 Mbps)
# mean service times (s)
s_v = (L_v * 8) / C_link
s_d = (L_d * 8) / C_link

# lambdas to evaluate (packets/s)
lambdas = [300, 325, 350, 375, 400, 425]
```

```python
def mean_response_times(lambda_val):
    # lambda_val: arrival rate per class
    Lambda = 2.0 * lambda_val
    E_S = 0.5 * (s_v + s_d)
    E_S2 = s_v**2 + s_d**2
    rho = Lambda * E_S
    if rho >= 1.0:
        return None, None  # unstable
    E_W = (Lambda * E_S2) / (2.0 * (1.0 - rho))
    T_v = E_W + s_v
    T_d = E_W + s_d
    return T_v, T_d
```

```python
T_v_list = []
T_d_list = []
for lam in lambdas:
    Tv, Td = mean_response_times(lam)
    if Tv is None:
        T_v_list.append(np.nan)
        T_d_list.append(np.nan)
    else:
        # Convert to milliseconds for readability
        T_v_list.append(Tv * 1000.0)
        T_d_list.append(Td * 1000.0)
```

```python
# Plotting
plt.figure(figsize=(8,5))
plt.plot(lambdas, T_v_list, marker='o', label='Video (L=120 B)')
plt.plot(lambdas, T_d_list, marker='s', label='Data  (L=135 B)')
plt.xlabel('Arrival rate per class λ (packets/s)')
plt.ylabel('Mean response time (ms)')
plt.title('Mean response time vs λ (per-class), Link=1 Mbps')
plt.grid(True, which='both', linestyle='--', alpha=0.6)
plt.legend()
plt.show()

for lam, tv, td in zip(lambdas, T_v_list, T_d_list):
    print(f"λ={lam:3d} pkt/s:  T_video={tv:7.4f} ms,  T_data={td:7.4f} ms")
```
